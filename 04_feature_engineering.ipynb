```python
#Importing the libraries
import pandas as pd
```

```python
# Loading the dataset
df = pd.read_csv("data/final_data.csv")
```

```python
# Checking the shape
df.shape
```

```python
# Checking the first 5 rows
df.head()
```

```python
# Creating new features based on the eda

# Balance per product
df['balance_per_product'] = df['balance'] / (df['num_of_products'] + 1)

# Engagement Score
df['engagement_score'] = df['num_of_products'] * df['is_active_member']

# Age group
# Create age groups
df['age_group'] = pd.cut(
    df['age'],
    bins=[18, 30, 45, 60, 100],
    labels=['(18, 30]', '(30, 45]', '(45, 60]', '(60, 100]'],
    include_lowest=True
).astype(str)

# FILL NaN VALUES
df['age_group'] = df['age_group'].fillna('(60, 100]')

# Balance Indicator
df['zero_balance'] = (df['balance'] == 0).astype(int)
```

```python
# Saving the dataset
df.to_csv('data/df.csv', index=False)
```
